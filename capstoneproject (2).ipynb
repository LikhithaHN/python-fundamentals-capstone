```python
import pandas as pd              # import pandas library for data manipulation             
import numpy as np               # import numpy for handling missing values
employee_data = {                # create dictionary with employee details
    'ID':     ['A001', 'A002','A003','A004','A005'],   # employee IDs
    'Name':   ['John Alter','Alice Luxumberg','Tom Sabestine','Nina Adgra','Amy Johny'],# names
    'Gender': ['M','F','M','F','F' ],           # gender M or F
    'City':   ['Paris','London','Berlin','Newyork','Madrid'],    # city locations  
    'Age':    [25,27,29,31,30]             # age in years
}
df_employee = pd.DataFrame(employee_data) # convert dictionary to DataFrame
print(df_employee)                        # display the DataFrame

```

```python
seniority_data = {                                        # create dictionary with seniority details
    'ID':                ['A001', 'A002','A003','A004','A005'], # employee IDs
    'Designation Level': [2,2,3,2,3]           # designation levels (1=highest, 4=lowest)
}
df_seniority = pd.DataFrame(seniority_data)   # convert dictionary to DataFrame
print(df_seniority)                            # display the DataFrame
```

```python
project_data = {                                # create dictionary with project details
     'ID':      ['A001','A002','A003','A004','A005', # employee IDs for each project
                'A002','A005','A003','A001','A003',
                'A001','A004','A004','A005'],
'Project':      ['Project 1','Project 2','Project 3','Project 4','Project 5', # project names
                'Project 6','Project 7','Project 8','Project 9','Project 10',
                'Project 11','Project 12','Project 13','Project 14'],
   'Cost':      [1002000, 2000000, 4500000, 5500000, np.nan,  # project costs (np.nan = missing)
                680000, 400000, 350000, np.nan, 300000,
                2000000, 1000000, 3000000, 200000],
 'Status':      ['Finished','Ongoing','Finished','Ongoing','Finished','Failed',  # project status
                 'Finished','Failed','Ongoing','Finished','Failed','Ongoing','Finished','Finished']   
}
df_project = pd.DataFrame(project_data) # convert dictionary to DataFrame
print(df_project)                       # display the DataFrame

```

```python
df_employee.to_csv('employee.csv', index=False) # save employee DataFrame to CSV
df_seniority.to_csv('seniority.csv', index=False) # save seniority DataFrame to CSV
df_project.to_csv('project.csv', index=False)      # save project DataFrame to CSV
print("All 3 CSV files saved successfully!")   # confirm all files saved
```

```python
df_project = pd.read_csv('project.csv')   # read project CSV 
running_sum   = 0                 # variable to sum cumulative  costs
running_count = 0                # variable to track count of known costs
for i in range(len(df_project)):  # loop through every row
     if pd.isna(df_project.loc[i, 'Cost']):       # check if cost is missing
        running_avg = running_sum / running_count  # calculate running average
        df_project.loc[i, 'Cost'] = running_avg   # fill missing value with average
     else:                                         # if cost is not missing
        running_sum   += df_project.loc[i, 'Cost'] # add cost to running sum
        running_count += 1                    # increment count       
df_project.to_csv('project.csv', index=False) # save updated CSV
pd.options.display.float_format = '{:.0f}'.format # display without scientific notation
print(df_project[['Project', 'Cost']])   # print Project and Cost columns
```

```python
df_employee = pd.read_csv('employee.csv') # read employee CSV saved in Task 1
if 'Name' in df_employee.columns:  # Only split if 'Name' column still exists
    df_employee['First Name'] = df_employee['Name'].str.split(' ').str[0] # split Name on space, take first word
    df_employee['Last Name']  = df_employee['Name'].str.split(' ').str[1]  # split Name on space, take second word
    df_employee.drop(columns=['Name'], inplace=True) # remove original Name column
print(df_employee) # display updated DataFrame
df_employee.to_csv('employee.csv', index=False) # save updated employee CSV
```

```python
df_employee  = pd.read_csv('employee.csv') # read updated employee CSV
df_seniority = pd.read_csv('seniority.csv') # read seniority CSV
df_project   = pd.read_csv('project.csv') # read updated project CSV
df_emp_sen = pd.merge(df_employee, df_seniority, on='ID', how='left') # merge employee and seniority on ID
Final = pd.merge(df_emp_sen, df_project, on='ID', how='left')   # merge result with project on ID
print(Final)   # display the Final merged DataFrame
```

```python
Final['Bonus'] = Final.apply( # add new Bonus column to Final DataFrame
    lambda row: row['Cost'] * 0.05 # give 5% of cost as bonus
    if row['Status'] == 'Finished' else 0,  # only for Finished projects, else 0
    axis=1  # apply function row by row
)
print(Final[['ID', 'First Name', 'Project', 'Cost', 'Status', 'Bonus']]) # display selected columns
```

```python
Final.loc[Final['Status'] == 'Failed', 'Designation Level'] += 1 # demote Failed projects by adding 1 to designation level
Final = Final[Final['Designation Level'] <= 4].reset_index(drop=True) # delete rows where designation level exceeds 4
print(Final[['ID', 'First Name', 'Project', 'Status', 'Designation Level']]) # display selected columns
```

```python
Final['First Name'] = Final.apply(    # update First Name column
    lambda row: ('Mr. ' if row['Gender'] == 'M' else 'Mrs. ') + row['First Name'],# add Mr. if gender is Male
    axis=1 # add Mrs. if gender is Female
) 
Final.drop(columns=['Gender'], inplace=True) # drop Gender column as it is no longer needed
print(Final[['ID', 'First Name', 'Last Name']]) # display ID, First Name, Last Name columns
```

```python

Final['Designation Level'] = Final.apply(   # update Designation Level column
    lambda row: row['Designation Level'] - 1  # promote by subtracting 1 from level 
    if row['Age'] > 29   # only for employees with age more than 29 
    else row['Designation Level'],  # keep same level if age is 29 or below
    axis=1  # apply function row by row
)
print(Final[['ID', 'First Name', 'Age', 'Designation Level']].drop_duplicates('ID'))# display selected columns
```

```python
TotalProjCost = Final.groupby (['ID', 'First Name'], as_index=False)['Cost'].sum() #group by ID&First Name,sum all project costs
TotalProjCost = TotalProjCost.rename(columns={'Cost': 'Total Cost'}) # rename Cost column to Total Cost
TotalProjCost = TotalProjCost[['ID', 'First Name', 'Total Cost']] # keep only 3 required columns
print(TotalProjCost) # display the TotalProjCost DataFrame
```

```python
city_with_o = Final[Final['City'].str.contains('o', case=False, na=False)]  # filter rows where City contains letter 'o' (case-insensitive)
print(city_with_o[['ID', 'First Name', 'Last Name', 'City', 'Age']].drop_duplicates('ID'))#employee details whose city name contains the letter “o” in it
```
